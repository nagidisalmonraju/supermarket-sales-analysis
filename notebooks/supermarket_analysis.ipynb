```python
import pandas as pd

# Load the cleaned supermarket sales dataset
df = pd.read_csv("../data/cleaned_supermarket_sales.csv")

# Check the dataset
print(df.columns.tolist())
print(df.head())
```

```python
import matplotlib.pyplot as plt
import calendar
import os

# Calculate monthly sales
monthly_sales = df.groupby("Month")["Total"].sum().sort_index()

# Convert numeric months to month names
monthly_sales.index = [
    calendar.month_name[int(month)]
    for month in monthly_sales.index
]

# Create the monthly sales trend chart
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.xticks(rotation=0)

# Save the chart in the images folder
os.makedirs("../images", exist_ok=True)

plt.tight_layout()
plt.savefig(
    "../images/monthly_sales_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("monthly_sales_trend.png saved successfully!")
```
